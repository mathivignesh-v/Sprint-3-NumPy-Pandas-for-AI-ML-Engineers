## CSV Files

### Step 1 – Understand the Concept

CSV is a delimited text format for tabular data.

### Step 2 – Demonstrate the Concept

- **Simple example:** It is widely used for simple data exchange.
- **Real-world analogy:** A plain-text table separated by commas.
- **Business example:** Exports from billing systems are usually CSV.
- **AI/ML example:** Most public ML datasets are CSV files.

Each example represents the concept because it shows the same idea: apply the operation to many values or records in one consistent step.

### Step 3 – Implement the Concept

````python
import pandas as pd
from io import StringIO
text='id,name,score\n1,Asha,90\n2,Bala,85\n'
print(pd.read_csv(StringIO(text)))
````

- **Expected output:**

```text

0   1  Asha     90
1   2  Bala     85
```
- **Where it is used in AI/ML or data analytics:** Most public ML datasets are CSV files.

### Interpretation

The text is parsed into columns and rows.

### Key Takeaway

**Key Takeaway:** The text is parsed into columns and rows.

## Pickle Files

### Step 1 – Understand the Concept

Pickle serializes Python objects. It is convenient in trusted Python environments but unsafe for untrusted input.

### Step 2 – Demonstrate the Concept

- **Simple example:** Use it only when the source is trusted.
- **Real-world analogy:** Freezing a Python object in a jar to reopen later.
- **Business example:** Saving an intermediate Python object between steps.
- **AI/ML example:** Trained models and preprocessors are often pickled.

Each example represents the concept because it shows the same idea: apply the operation to many values or records in one consistent step.

### Step 3 – Implement the Concept

````python
import pickle
obj={'name':'Asha','score':90}
with open('sample.pkl','wb') as f: pickle.dump(obj,f)
with open('sample.pkl','rb') as f: print(pickle.load(f))
````

### Code Explanation

- **What the code does and why it is used:** Dumps a dictionary to a pickle file and loads it back. Only load pickles you trust.
- **Expected output:**

```text
{'name': 'Asha', 'score': 90}
```
- **Where it is used in AI/ML or data analytics:** Trained models and preprocessors are often pickled.

### Interpretation

The Python object is restored. Never load untrusted pickle files.

### Key Takeaway

**Key Takeaway:** The Python object is restored.

## SQL Tables

### Step 1 – Understand the Concept

Pandas can load SQL query results into DataFrames.

### Step 2 – Demonstrate the Concept

- **Simple example:** Database-to-analysis workflows commonly use this pattern.
- **Real-world analogy:** Asking a database a question and getting a table.
- **Business example:** Reading customers from a company database.
- **AI/ML example:** Training data is frequently pulled from SQL databases.

Each example represents the concept because it shows the same idea: apply the operation to many values or records in one consistent step.

### Step 3 – Implement the Concept

````python
import sqlite3, pandas as pd
conn=sqlite3.connect(':memory:')
pd.DataFrame({'id':[1,2],'name':['Asha','Bala']}).to_sql('customers',conn,index=False)
print(pd.read_sql('SELECT * FROM customers WHERE id>=1',conn))
conn.close()
````

### Code Explanation

- **What the code does and why it is used:** Uses an in-memory SQLite database with to_sql and read_sql so no server is needed.
- **Expected output:**

```text
   id  name
0   1  Asha
1   2  Bala
```
- **Where it is used in AI/ML or data analytics:** Training data is frequently pulled from SQL databases.

### Interpretation

The SQL query result becomes a DataFrame.

### Key Takeaway

**Key Takeaway:** The SQL query result becomes a DataFrame.

## Reading Multiple Files

### Step 1 – Understand the Concept

Multiple compatible files can be read and concatenated.

### Step 2 – Demonstrate the Concept

- **Simple example:** Daily CSV files can be combined into one analytical dataset.
- **Real-world analogy:** Stacking monthly reports into one yearly report.
- **Business example:** Combining one CSV per month.
- **AI/ML example:** Training data is often split into many files.

Each example represents the concept because it shows the same idea: apply the operation to many values or records in one consistent step.

### Step 3 – Implement the Concept

````python
import pandas as pd
from pathlib import Path
Path('data_demo').mkdir(exist_ok=True)
pd.DataFrame({'m':['Jan'],'v':[1]}).to_csv('data_demo/jan.csv',index=False)
pd.DataFrame({'m':['Feb'],'v':[2]}).to_csv('data_demo/feb.csv',index=False)
files=sorted(Path('data_demo').glob('*.csv'))
print('files:',len(files))
print(pd.concat([pd.read_csv(f) for f in files],ignore_index=True))
````

### Code Explanation

- **What the code does and why it is used:** Creates two CSVs, finds them with glob, reads each and concatenates them.
- **Expected output:**

```text
files: 2
     m  v
0  Feb  2
1  Jan  1
```
- **Where it is used in AI/ML or data analytics:** Training data is often split into many files.

### Interpretation

The files are found, read and combined into one DataFrame.

### Key Takeaway

**Key Takeaway:** The files are found, read and combined into one DataFrame.

## Best Practices

- Use explicit file paths and verify that the expected input file exists.
- Choose the correct Pandas reader and writer for CSV, Excel, JSON, or other formats.
- Inspect row counts, column names, and dtypes after loading a file.
- Specify encoding when needed and handle parsing errors intentionally.
- Confirm the output destination and overwrite behavior before saving important data.

## Common Mistakes

- Using a relative path from an unexpected working directory.
- Reading a file with a function that does not match its format.
- Ignoring delimiters, encodings, date formats, or missing-value conventions.
- Overwriting the source file accidentally.
- Assuming a save succeeded without checking or reading back the output.

## Advantages

- Pandas provides convenient APIs for importing and exporting common data formats.
- File operations support repeatable data preparation and reporting workflows.
- Different formats can suit different needs, from portable CSV files to Excel workbooks.
- Separating source data from generated outputs improves reproducibility.

## Limitations

- Formats preserve different amounts of type and workbook metadata.
- Very large files may need chunked reading to manage memory.
- Paths, permissions, encodings, and dependencies can vary between systems.
- Successful loading does not guarantee that the data is complete or correct.
