## fillna()

### Step 3 – Implement the Concept

````python
import pandas as pd
df=pd.DataFrame({'score':[80,None,90]})
df['filled']=df['score'].fillna(df['score'].median())
print(df)
````

## Missing Value Handling Strategies

### Step 1 – Understand the Concept

Possible strategies include deletion, statistical imputation, forward/backward filling, interpolation, model-based imputation, or missing indicators. The reason for missingness matters.

### Step 2 – Demonstrate the Concept

- **Simple example:** For ML, fit imputation rules on training data only to avoid leakage.
- **Real-world analogy:** Choosing the right fix for each gap.
- **Business example:** Imputing and flagging missing values.
- **AI/ML example:** Fit imputers on training data only.

Each example represents the concept because it shows the same idea: apply the operation to many values or records in one consistent step.

### Step 3 – Implement the Concept

````python
import pandas as pd
df=pd.DataFrame({'age':[20,21,None,24],'income':[30000,32000,35000,None]})
df['age_missing']=df['age'].isna().astype(int)
df['age']=df['age'].fillna(df['age'].median())
df['income']=df['income'].fillna(df['income'].median())
print(df)
````

### Code Explanation

- **What the code does and why it is used:** Imputes with the median and adds a missing flag.
- **Expected output:**

```text
    age   income  age_missing
0  20.0  30000.0            0
1  21.0  32000.0            0
2  21.0  35000.0            1
3  24.0  32000.0            0
```
- **Where it is used in AI/ML or data analytics:** Fit imputers on training data only.

### Interpretation

The example imputes values and preserves a missingness indicator.

### Key Takeaway

**Key Takeaway:** The example imputes values and preserves a missingness indicator.

## Best Practices

- Measure missingness by column and row before choosing a strategy.
- Choose dropping or imputation based on the variable type, purpose, and reason values are missing.
- Use statistics such as median only when appropriate for the distribution and domain.
- Consider a missingness indicator if the absence of a value may itself be informative.
- For ML, fit imputation on training data and apply the learned rule to validation/test data to avoid leakage.

## Common Mistakes

- Replacing every missing value with zero when zero has a different meaning.
- Dropping all rows with any missing value and removing a large or biased portion of data.
- Imputing before the train/test split and allowing test data to influence preprocessing.
- Using mean or median without checking skew, outliers, or variable meaning.
- Ignoring empty strings or placeholder text such as 'unknown' that may represent missing data.

## Advantages

- Pandas offers isna(), notna(), dropna(), fillna(), and interpolate() for missing-data workflows.
- Missingness summaries help reveal data-collection issues.
- Suitable imputation can preserve useful records that deletion would remove.
- Documented handling rules make preprocessing more transparent and repeatable.

## Limitations

- No single strategy works for every dataset or missingness pattern.
- Imputation introduces estimates and may reduce variability or bias relationships.
- Deletion can reduce sample size and disproportionately remove groups.
- Interpolation is only suitable for certain ordered data and is not appropriate for every variable.

**Q5. What is imputation leakage?**

**Answer:** Using validation/test information to fit imputation rules, making evaluation overly optimistic.
