## Random Choice

````python
import numpy as np
rng=np.random.default_rng(1)
labels=np.array(['A','B','C'])
print(rng.choice(labels,8,p=[.5,.3,.2]))
````

## Random Distributions

### Step 1 – Understand the Concept

NumPy supports normal, uniform, binomial, Poisson, exponential, and other distributions.

### Step 2 – Demonstrate the Concept

- **Simple example:** Simulation should use a distribution that matches the process being modeled.
- **Real-world analogy:** Heights follow a bell curve; dice are uniform.
- **Business example:** Simulating demand.
- **AI/ML example:** Simulating data and noise.

Each example represents the concept because it shows the same idea: apply the operation to many values or records in one consistent step.

### Step 3 – Implement the Concept

````python
import numpy as np
rng=np.random.default_rng(10)
normal=rng.normal(50,10,1000)
uniform=rng.uniform(0,1,1000)
binomial=rng.binomial(10,.5,1000)
print(normal.mean(),uniform.mean(),binomial.mean())
````

### Code Explanation

- **What the code does and why it is used:** Draws normal, uniform and binomial samples and prints their means.
- **Expected output:**

```text
49.31110391424843 0.5025518756940434 4.96
```
- **Where it is used in AI/ML or data analytics:** Simulating data and noise.

### Interpretation

The sample statistics roughly reflect the chosen distributions.

### Key Takeaway

**Key Takeaway:** The sample statistics roughly reflect the chosen distributions.

## Best Practices

- Use np.random.default_rng(seed) when reproducible random experiments are required.
- Set and document a seed for repeatable demonstrations and tests.
- Choose a distribution and parameters that match the scenario being simulated.
- Check sample size, output shape, range, and summary statistics after generation.
- Keep the random generator state consistent when you need a repeatable sequence of draws.

## Common Mistakes

- Expecting identical results without controlling the random generator seed.
- Using the wrong distribution or parameters for the problem.
- Misunderstanding whether a random integer function includes its upper bound.
- Assuming a small random sample must match theoretical probabilities exactly.
- Resetting the generator repeatedly and unintentionally repeating the same sequence.

## Advantages

- NumPy generates large random samples efficiently.
- Generator seeds help make experiments reproducible.
- Many distributions support modeling different kinds of uncertainty.
- Random sampling is useful for simulations, synthetic data, and algorithm testing.

## Limitations

- Random samples can vary and may not represent a population well at small sizes.
- Results depend on choosing appropriate distributions and parameters.
- Reproducibility requires control of the generator and relevant execution details.
- NumPy pseudo-random generation is not designed for cryptographic security.
