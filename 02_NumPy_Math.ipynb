## Sum, Mean, Median

### Step 1 – Understand the Concept

Aggregation functions summarize data. Mean is the arithmetic average; median is the middle ordered value.

### Step 2 – Demonstrate the Concept

- **Simple example:** A salary dataset may use median when a few very high salaries distort the mean.
- **Real-world analogy:** Total bill, average bill and the middle bill when sorted.
- **Business example:** Average order value versus median order value when one huge order exists.
- **AI/ML example:** Mean and median summarise a feature and show skew caused by outliers.

Each example represents the concept because it shows the same idea: apply the operation to many values or records in one consistent step.

### Step 3 – Implement the Concept

````python
import numpy as np
x=np.array([10,20,20,30,100])
print(np.sum(x),np.mean(x),np.median(x))
````

### Code Explanation

- **What the code does and why it is used:** Computes sum, mean and median of an array containing one large value. The median resists outliers while the mean does not.
- **Expected output:**

```text
180 36.0 20.0
```
- **Where it is used in AI/ML or data analytics:** Mean and median summarise a feature and show skew caused by outliers.

### Interpretation

The extreme value affects the mean much more than the median.

### Key Takeaway

**Key Takeaway:** The extreme value affects the mean much more than the median.

## Variance and Standard Deviation

- **Real-world analogy:** How spread out the heights in a class are around the average height.

### Step 3 – Implement the Concept

**Key Takeaway:** They quantify dispersion.

## Matrix Multiplication

### Step 1 – Understand the Concept

Matrix multiplication combines rows and columns according to linear algebra rules.

### Step 2 – Demonstrate the Concept

- **Simple example:** Dense neural-network layers rely heavily on matrix multiplication.
- **Real-world analogy:** Combining two recipes' ingredient tables to get total ingredients per dish.

Each example represents the concept because it shows the same idea: apply the operation to many values or records in one consistent step.

### Step 3 – Implement the Concept

````python
import numpy as np
A=np.array([[1,2],[3,4]]); B=np.array([[5,6],[7,8]])
print(A@B)
print(np.matmul(A,B))
````

- **Expected output:**

```text

 [43 50]]
[[19 22]
 [43 50]]
```
- **Where it is used in AI/ML or data analytics:** Every dense neural network layer is a matrix multiplication.

**Key Takeaway:** Both expressions perform matrix multiplication.

## Basic Linear Algebra

### Step 1 – Understand the Concept

numpy.linalg provides operations such as determinant, norm, and solving linear systems.

### Step 2 – Demonstrate the Concept

- **Simple example:** These operations support mathematical modeling and optimization.
- **Real-world analogy:** Solving for the price of two items from two receipts.
- **Business example:** Solving simple allocation equations.
- **AI/ML example:** Linear regression and PCA rely on solving linear systems and norms.

Each example represents the concept because it shows the same idea: apply the operation to many values or records in one consistent step.

### Step 3 – Implement the Concept

````python
import numpy as np
A=np.array([[3.,1.],[1.,2.]])
b=np.array([9.,8.])
print(np.linalg.det(A))
print(np.linalg.norm(b))
print(np.linalg.solve(A,b))
````

### Code Explanation

- **What the code does and why it is used:** Computes a determinant, a norm and solves Ax=b. solve is preferred over computing an inverse.
- **Expected output:**

```text
5.000000000000001
12.041594578792296
[2. 3.]
```
- **Where it is used in AI/ML or data analytics:** Linear regression and PCA rely on solving linear systems and norms.

### Interpretation

The examples show determinant, vector norm, and a system solution.

### Key Takeaway

**Key Takeaway:** The examples show determinant, vector norm, and a system solution.

## Best Practices

- Check array shapes and dtypes before arithmetic or aggregation.
- Use NumPy functions such as np.sqrt(), np.mean(), np.sum(), and np.round() where appropriate.
- Understand broadcasting before combining arrays with different shapes.
- Check for zero denominators and invalid inputs to logarithms or square roots.
- Use np.isclose() for approximate comparisons of floating-point results.

## Common Mistakes

- Assuming all arrays of different shapes can be broadcast together.
- Using exact equality to compare floating-point values when a tolerance is needed.
- Allowing divide-by-zero or invalid mathematical operations to go unchecked.
- Confusing element-wise multiplication (*) with matrix multiplication (@).
- Rounding intermediate values too early and losing precision.

## Advantages

- Vectorized mathematical functions apply operations to whole arrays efficiently.
- Broadcasting supports calculations between scalars and compatible arrays.
- Aggregation functions summarize numerical data with concise expressions.
- NumPy math operations are useful for statistics, simulations, and feature preparation.

## Limitations

- Floating-point calculations may contain small representation errors.
- Broadcasting can hide unintended shape mismatches if outputs are not inspected.
- Invalid inputs may produce warnings, NaN, or infinity.
- NumPy cannot determine whether a mathematically valid result makes sense in a particular business context.
